# Reducir datos

El dato original tiene 1.5GB por archivo, y tiene muchos columnas no nos interesadas.

Por lo tanto lo decidimos reducir.

Datos originales: [matrices-de-viaje](https://www.dtpm.cl/index.php/documentos/matrices-de-viaje) de dtpm

```python
import pandas as pd
```

```python
etapas = [
    "../data/etapas/2025-04-21.etapas.csv",
    "../data/etapas/2025-04-22.etapas.csv",
    "../data/etapas/2025-04-23.etapas.csv",
    "../data/etapas/2025-04-27.etapas.csv",
    "../data/etapas/2025-04-24.etapas.csv",
    "../data/etapas/2025-04-25.etapas.csv",
    "../data/etapas/2025-04-26.etapas.csv"
]

dfs = []
for etapa in etapas:
    df = pd.read_csv(etapa, delimiter="|")
    dfs.append(df)
```

Las columnas interesadas son: `tipo_transporte`, `tiene_bajada`, `tiempo_subida`, `tiempo_bajada`, `tiempo_etapa`, `comuna_subida`, `comuna_bajada`, `parada_subida`, `parada_bajada`, `dist_ruta_paraderos`, `dist_eucl_paraderos`, `x_subida`,`y_subida`, `x_bajada`, `y_bajada`

```python
COLS = [
    "tipo_transporte", "tiene_bajada",
    "tiempo_subida", "tiempo_bajada", "tiempo_etapa",
    "comuna_subida", "comuna_bajada",
    "parada_subida", "parada_bajada",
    "dist_ruta_paraderos", "dist_eucl_paraderos",
    "x_subida","y_subida", "x_bajada", "y_bajada"
]

reduced_dfs = [
    df[COLS] for df in dfs
]
```

```python
# guardar
for i, df in enumerate(reduced_dfs):
    date = reduced_dfs[i].loc[0, "tiempo_subida"][:10]

    # reduced
    df.to_csv(f"../data/reduced/{date}.csv", index=False)

    # sample
    df[:5000].to_csv(f"../data/sample/{date}.csv", index=False)
```
